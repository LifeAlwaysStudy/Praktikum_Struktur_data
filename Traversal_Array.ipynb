```python
# 5. Tiga cara menelusuri list

nilai = [78, 85, 90, 75, 88]

# 1) Traversal berdasarkan nilai: cocok jika hanya butuh nilainya
for x in nilai:
    print(x)
```

```python
# 2) Traversal berdasarkan indeks: cocok jika butuh kontrol posisi
for i in range(len(nilai)):
    print(i, nilai[i])
```

```python
# 3) enumerate(): indeks dan nilai sekaligus, lebih rapi dibaca
for i, x in enumerate(nilai):
    print('Indeks', i, 'nilai', x)
```
