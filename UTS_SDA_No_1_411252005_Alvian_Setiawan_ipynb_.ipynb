```python
NIM = "12345677"   # <-- ganti dengan NIM kamu
d = int(NIM[-1])

data = [
    ['TI001', 'Alya', 78], ['TI002', 'Bima', 80 + d], ['TI003', 'Citra', 67],
    ['TI004', 'Danu', 92], ['TI005', 'Eka', 74],      ['TI006', 'Fajar', 88],
    ['TI007', 'Gina', 81], ['TI008', 'Hadi', 65 + d], ['TI009', 'Intan', 91],
    ['TI010', 'Joko', 76], ['TI011', 'Karin', 84],    ['TI012', 'Lukman', 70 + d]
]

def tampil(judul, daftar):
    print(judul)
    for baris in daftar:
        print("  ", baris)

# 1. Linear Search manual
def linear_search_nim(data, target):
    perbandingan = 0
    for i in range(len(data)):
        perbandingan += 1
        if data[i][0] == target:
            return i, perbandingan
    return -1, perbandingan

# 2. Uji Linear Search
target_index = d % len(data)
target_nim = data[target_index][0]
idx_lin, cmp_lin = linear_search_nim(data, target_nim)
idx_lin_x, cmp_lin_x = linear_search_nim(data, "TI999")
print(f"target_index={target_index}, target_nim={target_nim}")
print(f"Linear '{target_nim}' -> indeks {idx_lin}, perbandingan {cmp_lin}")
print(f"Linear 'TI999'  -> indeks {idx_lin_x}, perbandingan {cmp_lin_x}")

# 3. Merge dan Merge Sort rekursif
def merge(left, right):
    hasil, i, j = [], 0, 0
    while i < len(left) and j < len(right):
        if left[i][0] <= right[j][0]:
            hasil.append(left[i]); i += 1
        else:
            hasil.append(right[j]); j += 1
    while i < len(left):
        hasil.append(left[i]); i += 1
    while j < len(right):
        hasil.append(right[j]); j += 1
    return hasil

def merge_sort_nim(data):
    if len(data) <= 1:                 # base case
        return data[:]
    mid = len(data) // 2
    return merge(merge_sort_nim(data[:mid]), merge_sort_nim(data[mid:]))

# 4. Sebelum dan sesudah sorting (data asli tidak diubah)
data_awal = [b[:] for b in data]
tampil("Data SEBELUM diurutkan:", data)
data_urut = merge_sort_nim(data)
tampil("Data SESUDAH diurutkan:", data_urut)
print("Data awal tidak berubah:", data == data_awal)
print("Uji data dibalik -> benar:", merge_sort_nim(data[::-1]) == data_urut)

# 5. Binary Search manual
def binary_search_nim(data_urut, target):
    low, high, perbandingan = 0, len(data_urut) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        perbandingan += 1
        if data_urut[mid][0] == target:
            return mid, perbandingan
        elif data_urut[mid][0] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, perbandingan

# 6. Uji Binary Search
idx_bin, cmp_bin = binary_search_nim(data_urut, target_nim)
idx_bin_x, cmp_bin_x = binary_search_nim(data_urut, "TI999")
print(f"Binary '{target_nim}' -> indeks {idx_bin}, perbandingan {cmp_bin}")
print(f"Binary 'TI999'  -> indeks {idx_bin_x}, perbandingan {cmp_bin_x}")

# 7. Trace Binary Search (opsi b)
def binary_search_trace(data_urut, target):
    low, high, langkah = 0, len(data_urut) - 1, 1
    while low <= high:
        mid = (low + high) // 2
        nilai = data_urut[mid][0]
        print(f"Langkah {langkah}: low={low}, high={high}, mid={mid}, data[mid]={nilai}", end=" -> ")
        if nilai == target:
            print("DITEMUKAN"); return mid
        elif nilai < target:
            print(f"geser low = {mid + 1}"); low = mid + 1
        else:
            print(f"geser high = {mid - 1}"); high = mid - 1
        langkah += 1
    print("Tidak ditemukan"); return -1

print("\nTrace Binary Search:", target_nim)
binary_search_trace(data_urut, target_nim)

# 8. Ringkasan
print("\n" + "=" * 50 + "\nRINGKASAN SOAL 1\n" + "=" * 50)
print("Target                     :", target_nim)
print("Indeks hasil (Linear/Binary):", idx_lin, "/", idx_bin)
print("Perbandingan Linear Search :", cmp_lin)
print("Perbandingan Binary Search :", cmp_bin)
print(f"TI999 -> Linear {cmp_lin_x}, Binary {cmp_bin_x} perbandingan")
tampil("Data akhir terurut:", data_urut)
```
