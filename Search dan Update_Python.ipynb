```python
    def search(self, target):
        current = self.head
        position = 0                     # penghitung posisi (linked list tidak punya indeks)
        while current is not None:
            if current.data == target:
                return position          # ketemu -> kembalikan posisinya
            current = current.next
            position += 1
        return -1                        # tidak ditemukan
```
