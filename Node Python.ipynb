```python
# Membuat class Node, membuat object, dan menghubungkannya

# Cetak biru sebuah node
class Node:
    def __init__(self, data):
        self.data = data    # menyimpan nilai
        self.next = None    # belum menunjuk ke node lain

node_a = Node(10)
node_b = Node(20)
print(node_a.data)   # 10
print(node_a.next)   # None -> belum terhubung

# Menghubungkan: next milik node_a menunjuk ke object node_b
node_a.next = node_b
print(node_a.next.data)   # 20
```
