```python
# Fungsi Binary Search iteratif

def binary_search(data, target):
    left = 0                          # batas kiri ruang pencarian
    right = len(data) - 1             # batas kanan (inklusif)
    while left <= right:              # selama ruang pencarian masih ada
        mid = (left + right) // 2     # indeks tengah
        if data[mid] == target:
            return mid                # ketemu
        elif data[mid] < target:
            left = mid + 1            # target di kanan, buang bagian kiri
        else:
            right = mid - 1           # target di kiri, buang bagian kanan
    return -1                         # ruang habis, target tidak ada

data = [12, 18, 25, 31, 44, 57, 63, 70, 82]   # HARUS terurut
print(binary_search(data, 57))   # 5
print(binary_search(data, 13))   # -1
```

```python
# Binary Search dengan trace

def binary_search_trace(data, target):
    left = 0
    right = len(data) - 1
    while left <= right:
        mid = (left + right) // 2
        # cetak kondisi batas di setiap langkah
        print('left=', left, 'right=', right, 'mid=', mid, 'nilai=', data[mid])
        if data[mid] == target:
            return mid
        elif data[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

binary_search_trace([12, 18, 25, 31, 44, 57, 63, 70, 82], 57)
```
