```python
NIM = "12345677"   # <-- ganti dengan NIM kamu
d = int(NIM[-1])

# ===== Struktur data =====
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedQueue:
    """Queue FIFO berbasis singly linked list (front, rear, count)."""
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0

    def is_empty(self):
        return self.count == 0

    def size(self):
        return self.count

    def enqueue(self, data):               # tambah di belakang, O(1)
        node = Node(data)
        if self.is_empty():
            self.front = self.rear = node
        else:
            self.rear.next = node
            self.rear = node
        self.count += 1

    def enqueue_front(self, data):         # tambah di depan (untuk UNDO), O(1)
        node = Node(data)
        node.next = self.front
        self.front = node
        if self.rear is None:              # queue sebelumnya kosong
            self.rear = node
        self.count += 1

    def dequeue(self):                     # ambil dari depan, O(1)
        if self.is_empty():
            print("[PESAN] Queue kosong, tidak ada yang bisa dilayani (dequeue dibatalkan).")
            return None
        node = self.front
        self.front = node.next
        if self.front is None:             # queue jadi kosong
            self.rear = None
        self.count -= 1
        return node.data

    def peek(self):
        if self.is_empty():
            print("[PESAN] Queue kosong, tidak ada elemen front.")
            return None
        return self.front.data

    def display(self):
        isi, cur = [], self.front
        while cur is not None:
            isi.append(cur.data)
            cur = cur.next
        print("Queue (front -> rear):", " -> ".join(isi) if isi else "(kosong)")

class Stack:
    """Stack LIFO dengan Python list sebagai penyimpanan internal."""
    def __init__(self):
        self._items = []

    def push(self, data):
        self._items.append(data)

    def pop(self):
        if self.is_empty():
            print("[PESAN] Stack kosong, tidak ada riwayat untuk di-pop.")
            return None
        return self._items.pop()

    def peek(self):
        if self.is_empty():
            print("[PESAN] Stack kosong, tidak ada elemen top.")
            return None
        return self._items[-1]

    def is_empty(self):
        return len(self._items) == 0

    def size(self):
        return len(self._items)

    def display(self):
        print("Stack (bawah -> top):", " | ".join(self._items) if self._items else "(kosong)")
```

```python
# ===== Tugas 1: uji operasi dasar =====
print("--- Uji LinkedQueue ---")
q = LinkedQueue()
print("is_empty:", q.is_empty(), "| size:", q.size())
q.enqueue("X1"); q.enqueue("X2"); q.enqueue("X3"); q.display()
print("peek:", q.peek())
q.enqueue_front("X0"); q.display()
print("dequeue:", q.dequeue()); q.display()
print("size:", q.size(), "| front:", q.front.data, "| rear:", q.rear.data)

print("\n--- Uji Stack ---")
s = Stack()
print("is_empty:", s.is_empty(), "| size:", s.size())
s.push("Y1"); s.push("Y2"); s.push("Y3"); s.display()
print("peek:", s.peek())
print("pop:", s.pop()); s.display()
print("size:", s.size())
```

```python
# ===== Tugas 2: simulasi 12 operasi =====
A = f'MHS-{d}01'; B = f'MHS-{d}02'; C = f'MHS-{d}03'
D = f'MHS-{d}04'; E = f'MHS-{d}05'; F = f'MHS-{d}06'

antrean = LinkedQueue()
riwayat = Stack()

def layani():
    x = antrean.dequeue()
    if x is not None:
        riwayat.push(x)
    return x

def undo():
    x = riwayat.pop()
    if x is not None:
        antrean.enqueue_front(x)
    return x

operasi = [("DATANG", A), ("DATANG", B), ("DATANG", C), ("LAYANI", None),
           ("DATANG", D), ("LAYANI", None), ("DATANG", E), ("UNDO", None),
           ("LAYANI", None), ("LAYANI", None), ("UNDO", None), ("DATANG", F)]

for no, (op, x) in enumerate(operasi, start=1):
    if op == "DATANG":
        antrean.enqueue(x); ket = f"DATANG {x}"
    elif op == "LAYANI":
        ket = f"LAYANI ({layani()})"
    else:
        ket = f"UNDO ({undo()})"
    print(f"Langkah {no}: {ket}")
    antrean.display()
    riwayat.display()
    print()

# ===== Tugas 3: kondisi akhir =====
print("Front Queue  :", antrean.front.data if antrean.front else None)
print("Rear Queue   :", antrean.rear.data if antrean.rear else None)
print("Jumlah Queue :", antrean.size())
print("Jumlah Stack :", riwayat.size())
```
