```python
NIM = "411252005"   # GANTI dengan NIM Anda
d = int(NIM[-1])
data = [
    ['TI001', 'Alya',   78],
    ['TI002', 'Bima',   80 + d],
    ['TI003', 'Citra',  67],
    ['TI004', 'Danu',   92],
    ['TI005', 'Eka',    74],
    ['TI006', 'Fajar',  88],
    ['TI007', 'Gina',   81],
    ['TI008', 'Hadi',   65 + d],
    ['TI009', 'Intan',  91],
    ['TI010', 'Joko',   76],
    ['TI011', 'Karin',  84],
    ['TI012', 'Lukman', 70 + d],
]
data_backup = [row[:] for row in data]   # untuk membuktikan data awal tidak berubah
```

```python
def linear_search_nim(data, target):
    comparisons = 0
    for i in range(len(data)):
        comparisons += 1
        if data[i][0] == target:
            return i, comparisons
    return -1, comparisons


def merge(left, right, counter=None):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if counter is not None:
            counter[0] += 1
        if left[i][0] <= right[j][0]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    while i < len(left):
        result.append(left[i]); i += 1
    while j < len(right):
        result.append(right[j]); j += 1
    return result


def merge_sort_nim(data, counter=None, trace=None, depth=0):
    if len(data) <= 1:                      # base case rekursi
        return data[:]
    mid = len(data) // 2
    left_part, right_part = data[:mid], data[mid:]
    if trace is not None:
        trace.append(f"{'  '*depth}SPLIT  {[x[0] for x in data]} -> "
                     f"{[x[0] for x in left_part]} | {[x[0] for x in right_part]}")
    left = merge_sort_nim(left_part, counter, trace, depth + 1)    # rekursi kiri
    right = merge_sort_nim(right_part, counter, trace, depth + 1)  # rekursi kanan
    merged = merge(left, right, counter)
    if trace is not None:
        trace.append(f"{'  '*depth}MERGE  {[x[0] for x in left]} + "
                     f"{[x[0] for x in right]} -> {[x[0] for x in merged]}")
    return merged


def binary_search_nim(data_urut, target):
    low, high = 0, len(data_urut) - 1
    comparisons = 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1                    # perbandingan kesamaan
        if data_urut[mid][0] == target:
            return mid, comparisons
        comparisons += 1                    # perbandingan lebih kecil/besar
        if data_urut[mid][0] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons
```

```python
target_index = d % len(data)
target_nim = data[target_index][0]
tidak_ada = "TI999"
print("d =", d, "| target_index =", target_index, "| target_nim =", target_nim)

idx_l, cmp_l = linear_search_nim(data, target_nim)
print(f"Linear {target_nim}: indeks={idx_l}, perbandingan={cmp_l}")
idx_l2, cmp_l2 = linear_search_nim(data, tidak_ada)
print(f"Linear {tidak_ada}: indeks={idx_l2}, perbandingan={cmp_l2}")
```

```python
print("Data SEBELUM diurutkan:")
for r in data: print("  ", r)

salinan = [row[:] for row in data]      # salinan, data awal tidak diubah
cnt = [0]
data_urut = merge_sort_nim(salinan, cnt)

print("\nData SESUDAH diurutkan:")
for r in data_urut: print("  ", r)
print("\nData awal tidak berubah:", data == data_backup)
print("Jumlah perbandingan Merge Sort:", cnt[0])
```

```python
idx_b, cmp_b = binary_search_nim(data_urut, target_nim)
print(f"Binary {target_nim}: indeks={idx_b}, perbandingan={cmp_b}")
idx_b2, cmp_b2 = binary_search_nim(data_urut, tidak_ada)
print(f"Binary {tidak_ada}: indeks={idx_b2}, perbandingan={cmp_b2}")
```

```python
print("=== TRACE MERGE SORT (6 data pertama) ===")
trace = []
merge_sort_nim([row[:] for row in data[:6]], None, trace)
print("\n".join(trace))

print("\n=== TRACE BINARY SEARCH untuk", target_nim, "===")
low, high, step = 0, len(data_urut) - 1, 1
while low <= high:
    mid = (low + high) // 2
    kode = data_urut[mid][0]
    print(f"Langkah {step}: low={low}, high={high}, mid={mid}, data[mid]={kode}", end=" -> ")
    if kode == target_nim:
        print("DITEMUKAN"); break
    elif kode < target_nim:
        print("lebih kecil, low = mid+1"); low = mid + 1
    else:
        print("lebih besar, high = mid-1"); high = mid - 1
    step += 1
```
