```python
# 1. Membuat list dan mengecek batas indeks

# Membuat list sebagai representasi array
nilai = [78, 85, 90, 67, 88]
print(nilai)
print(type(nilai))

# len() menghitung jumlah elemen; indeks terakhir = len - 1
print('Jumlah elemen:', len(nilai))
print('Indeks terakhir:', len(nilai) - 1)
print('Elemen terakhir:', nilai[len(nilai) - 1])
```

```python
# 2. Akses dan update elemen

nilai = [78, 85, 90, 67, 88]

# Akses berdasarkan indeks, kompleksitas O(1)
print(nilai[0])
print(nilai[2])
print(nilai[-1])

# Update elemen pada indeks tertentu, O(1), tidak ada elemen bergeser
nilai[3] = 75
print(nilai)
```

```python
# 3. Menambah dan menyisipkan elemen

# append(): tambah di akhir, O(1) amortized
nilai.append(92)
print(nilai)
# insert(i, x): sisip di indeks i, O(n) karena elemen sesudahnya bergeser ke kanan
nilai.insert(2, 80)
print(nilai)
```

```python
# 4. Menghapus elemen: pop, remove, del
# pop(i): hapus berdasarkan indeks dan kembalikan nilainya, O(n)
data = [10, 20, 30, 40, 50]
dihapus = data.pop(2)
print('Dihapus:', dihapus)
print('Data:', data)

# remove(x): hapus berdasarkan nilai (hanya kemunculan pertama)
data = [10, 20, 30, 20, 40]
data.remove(20)
print(data)

# del: hapus satu indeks atau rentang (slice, akhir eksklusif)
data = [10, 20, 30, 40, 50]
del data[1]
print(data)
del data[1:3]
print(data)
```
