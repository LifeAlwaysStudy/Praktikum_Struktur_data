```python
# Fungsi Linear Search
def linear_search(data, target):
    # periksa elemen satu per satu dari awal
    for index, value in enumerate(data):
        if value == target:
            return index
    return -1

data = [12, 35, 21, 48, 57, 63]
print('48 berada pada indeks:', linear_search(data, 48))   # 3
print('99 berada pada indeks:', linear_search(data, 99))   # -1
```

```python
# Mencari semua kemunculan
def linear_search_all(data, target):
    posisi = []                       # menampung semua indeks yang cocok
    for index, value in enumerate(data):
        if value == target:
            posisi.append(index)      # tidak berhenti, lanjut sampai akhir
    return posisi

print(linear_search_all([20, 10, 20, 30, 20], 20))   # [0, 2, 4]
```

```python
# Linear Search dengan jejak eksekusi (Contoh 2.16)

def linear_search_trace(data, target):
    for index, value in enumerate(data):
        print('Periksa indeks', index, 'nilai', value)
        if value == target:
            print('Target ditemukan')
            return index
    print('Target tidak ditemukan')
    return -1

linear_search_trace([7, 11, 15, 18, 25, 30], 25)
```
