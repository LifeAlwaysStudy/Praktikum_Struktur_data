```python
# Praktek Program Python Mencari NIM
nim = ['41250017', '41250003', '41250029', '41250011', '41250008']
target = '41250011'

# Data belum terurut -> pakai Linear Search
posisi = linear_search(nim, target)
if posisi != -1:
    print('NIM ditemukan pada indeks', posisi)   # indeks 3
else:
    print('NIM tidak ditemukan')

# Untuk banyak pencarian: urutkan dulu, lalu Binary Search
nim_urut = sorted(nim)       # sorted() membuat list baru, data asli tetap
print(nim_urut)
posisi = binary_search(nim_urut, target)
print('Posisi pada data terurut:', posisi)   # indeks mengacu ke nim_urut
```

```python
# Kesalahan Umum
# 14. IndexError dan Binary Search pada data tidak terurut
# Kesalahan 1: indeks di luar batas -> IndexError
data = [10, 20, 30]
print(data[3])   # error! indeks valid hanya 0, 1, 2

# Kesalahan 2: Binary Search pada data belum terurut -> hasil tidak bisa dipercaya
data = [40, 10, 30, 20, 50]
print(binary_search(data, 20))   # asumsi data terurut dilanggar
```
