```python
    # Hapus node pertama
    def delete_first(self):
        if self.head is None:          # edge case: list kosong
            return False
        self.head = self.head.next     # head loncat ke node kedua
        return True
```

```python
# Hapus berdasarkan nilai
    def delete(self, target):
        if self.head is None:                  # list kosong
            return False
        if self.head.data == target:           # target ada di head
            self.head = self.head.next
            return True
        previous = self.head                   # node sebelum current
        current = self.head.next
        while current is not None:
            if current.data == target:
                previous.next = current.next   # lewati (bypass) node target
                return True
            previous = current                 # geser kedua penunjuk bersama
            current = current.next
        return False                           # target tidak ada
```
