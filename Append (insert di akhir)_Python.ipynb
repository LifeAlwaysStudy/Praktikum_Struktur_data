```python
    def append(self, data):
        new_node = Node(data)
        if self.head is None:            # kasus khusus: list kosong
            self.head = new_node
            return                       # selesai, jangan lanjut ke traversal
        current = self.head
        while current.next is not None:  # cari node terakhir (next-nya None)
            current = current.next
        current.next = new_node          # sambungkan node baru di ujung
```
