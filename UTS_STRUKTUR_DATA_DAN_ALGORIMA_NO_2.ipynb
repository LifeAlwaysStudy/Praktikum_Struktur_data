```python
NIM = "411252005"
d = int(NIM[-1])
A = f'MHS-{d}01'; B = f'MHS-{d}02'; C = f'MHS-{d}03'
D = f'MHS-{d}04'; E = f'MHS-{d}05'; F = f'MHS-{d}06'


class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedQueue:
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0

    def is_empty(self):
        return self.count == 0

    def size(self):
        return self.count

    def enqueue(self, data):                 # tambah di belakang (rear)
        node = Node(data)
        if self.is_empty():
            self.front = self.rear = node
        else:
            self.rear.next = node
            self.rear = node
        self.count += 1

    def enqueue_front(self, data):           # tambah di depan (front), dipakai UNDO
        node = Node(data)
        if self.is_empty():
            self.front = self.rear = node
        else:
            node.next = self.front
            self.front = node
        self.count += 1

    def dequeue(self):
        if self.is_empty():
            print("  [Peringatan] Queue kosong, dequeue tidak dapat dilakukan.")
            return None
        data = self.front.data
        self.front = self.front.next
        self.count -= 1
        if self.front is None:               # queue menjadi kosong
            self.rear = None
        return data

    def peek(self):
        if self.is_empty():
            print("  [Peringatan] Queue kosong, tidak ada elemen front.")
            return None
        return self.front.data

    def display(self):
        items = []
        cur = self.front
        while cur is not None:
            items.append(cur.data)
            cur = cur.next
        return "front -> " + (" -> ".join(items) if items else "(kosong)") + " <- rear"


class Stack:
    def __init__(self):
        self._items = []

    def push(self, data):
        self._items.append(data)

    def pop(self):
        if self.is_empty():
            print("  [Peringatan] Stack kosong, pop tidak dapat dilakukan.")
            return None
        return self._items.pop()

    def peek(self):
        if self.is_empty():
            print("  [Peringatan] Stack kosong, tidak ada elemen top.")
            return None
        return self._items[-1]

    def is_empty(self):
        return len(self._items) == 0

    def size(self):
        return len(self._items)

    def display(self):
        return "bottom -> " + (" -> ".join(self._items) if self._items else "(kosong)") + " <- top"
```

```python
print("=== UJI DASAR LinkedQueue ===")
q = LinkedQueue()
q.enqueue("X1"); q.enqueue("X2"); q.enqueue_front("X0")
print(q.display(), "| size =", q.size(), "| peek =", q.peek())
print("dequeue ->", q.dequeue(), "|", q.display())
q.dequeue(); q.dequeue()
print("is_empty:", q.is_empty(), "| rear setelah kosong:", q.rear)

print("\n=== UJI DASAR Stack ===")
s = Stack()
s.push("S1"); s.push("S2"); s.push("S3")
print(s.display(), "| size =", s.size(), "| peek =", s.peek())
print("pop ->", s.pop(), "|", s.display())
s.pop(); s.pop()
print("is_empty:", s.is_empty())
```

```python
queue = LinkedQueue()
history = Stack()

def tampil(langkah, nama):
    print(f"\nLangkah {langkah}: {nama}")
    print("  Queue:", queue.display())
    print("  Stack:", history.display())

def datang(x):
    queue.enqueue(x)

def layani():
    served = queue.dequeue()
    if served is not None:
        history.push(served)
    return served

def undo():
    last = history.pop()
    if last is not None:
        queue.enqueue_front(last)
    return last

operasi = [
    ("DATANG A", lambda: datang(A)),
    ("DATANG B", lambda: datang(B)),
    ("DATANG C", lambda: datang(C)),
    ("LAYANI",   layani),
    ("DATANG D", lambda: datang(D)),
    ("LAYANI",   layani),
    ("DATANG E", lambda: datang(E)),
    ("UNDO",     undo),
    ("LAYANI",   layani),
    ("LAYANI",   layani),
    ("UNDO",     undo),
    ("DATANG F", lambda: datang(F)),
]

for i, (nama, aksi) in enumerate(operasi, start=1):
    aksi()
    tampil(i, nama)
```
